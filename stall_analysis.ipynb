## Introduction

This project analyses the sales and stock performance of Wanjiku's fresh-produce stall at Gikomba Market using core Python. The analysis covers daily sales performance, weekly growth, sales patterns, stock requirements and product profitability. The findings will be used to develop practical recommendations to improve business performance and inventory management.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall Sales Performance

This task examines the overall sales performance over the 14-day period by calculating total sales, average daily sales, highest daily sales and lowest daily sales. It also identifies the specific days and weeks in which the highest and lowest sales occurred using Python's built-in functions and list indexing.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]

best_week = best_index // 7 + 1
worst_week = worst_index // 7 + 1

print("OVERALL SALES PERFORMANCE")
print("-------------------------")
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f} on {best_day}, Week {best_week}")
print(f"Lowest sales: KES {lowest_sales:,.2f} on {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales Analysis

This task calculates the total sales for each week using Python list slicing. A function is developed to return the weekly totals as a tuple. The percentage change between the two weeks is then calculated to establish whether sales increased or decreased.

```python
week1 = daily_sales[0:7]
week2 = daily_sales[7:14]

print(week1)
print(week2)
```

```python
def week_totals(sales):
    """Calculate and return the sales totals for two weeks."""

    week1_total = sum(sales[0:7])
    week2_total = sum(sales[7:14])

    return week1_total, week2_total
```

```python
weekly_results = week_totals(daily_sales)

print(weekly_results)
```

```python
week1_total, week2_total = weekly_results

growth = ((week2_total - week1_total) / week1_total) * 100

print(f"Week 1 sales: KES {week1_total:,.2f}")
print(f"Week 2 sales: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth:.2f}%")
```

## Task 3: Daily Sales Classification

This task classifies daily sales into Strong, Normal and Weak categories based on their deviation from the overall average sales. A function is developed to classify each day's sales, and a dictionary is used to count the number of days in each category.

```python
print(daily_sales)
print(average_sales)
```

```python
strong_threshold = average_sales * 1.2
weak_threshold = average_sales * 0.8

print(f"Strong threshold: KES {strong_threshold:,.2f}")
print(f"Weak threshold: KES {weak_threshold:,.2f}")
```

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal or Weak."""

    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"
```

```python
print(classify_day(5200, average_sales))
print(classify_day(1950, average_sales))
print(classify_day(3600, average_sales))
```

```python
classifications = []

for sale in daily_sales:
    result = classify_day(sale, average_sales)
    classifications.append(result)

print(classifications)
```

```python
sales_count = {
    "Strong": classifications.count("Strong"),
    "Normal": classifications.count("Normal"),
    "Weak": classifications.count("Weak")
}

print(sales_count)
```

**Interpretation:** Out of the 14 days analysed, 4 days recorded strong sales, 4 days recorded normal sales and 6 days recorded weak sales. Weak sales were the most frequent category, suggesting a need to investigate factors contributing to lower daily sales.

## Task 4: Weekday versus Weekend Sales Analysis

This task compares average daily sales on weekdays (Monday to Friday) and weekends (Saturday and Sunday). The analysis uses list indexing and slicing to separate the two periods and calculates their respective average sales to identify differences in sales performance.

```python
weekday_sales = daily_sales[0:5] + daily_sales[7:12]

weekend_sales = daily_sales[5:7] + daily_sales[12:14]

print("Weekday sales:", weekday_sales)
print("Weekend sales:", weekend_sales)
```

```python
weekday_average = sum(weekday_sales) / len(weekday_sales)

weekend_average = sum(weekend_sales) / len(weekend_sales)

print(f"Average weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")
```

```python
percentage_difference = ((weekend_average - weekday_average) / weekday_average) * 100

print(f"Weekend sales are {percentage_difference:.2f}% higher than weekday sales.")
```

**Interpretation:** Average weekend sales were KES 4,900.00 compared with KES 2,685.00 on weekdays. Weekend sales were therefore 82.50% higher than weekday sales, indicating that the stall performed better on Saturdays and Sundays during the 14-day period.

## Task 5: Stock Management and Gross Profit Margins

This task examines the stall's stock levels to identify products that need restocking. A function is developed to compare the current quantity of each product with its reorder level. The gross profit margin is also calculated for each product to assess its profitability.

```python
print(stock)
```

```python
def products_to_reorder(stock):
    """Identify products that need restocking."""

    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list
```

```python
reorder_items = products_to_reorder(stock)

print("Products requiring restocking:", reorder_items)
```

```python
tomato_margin = ((stock["Tomatoes"]["price"] - stock["Tomatoes"]["cost"]) / stock["Tomatoes"]["price"]) * 100

print(f"Tomato gross profit margin: {tomato_margin:.2f}%")
```

```python
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100

    print(f"{product}: {margin:.2f}%")
```

**Interpretation:** Tomatoes and Sukuma require restocking because their current stock quantities are below their respective reorder levels. Avocado recorded the highest gross profit margin at 42.86%, followed by Sukuma at 40.00%, while Onions recorded the lowest margin at 30.00%. These findings can help Wanjiku make informed stock purchasing and pricing decisions.
